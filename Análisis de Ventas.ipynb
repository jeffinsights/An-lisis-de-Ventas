```python
import pandas as pd
```

```python

# Creamos un DataFrame de ventas
ventas_mensual = pd.DataFrame({'Mes': ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio'],'Ventas': [10000, 12000, 15000, 13000, 17000, 20000]})

ventas_mensual
```

```python
total_ventas=ventas_mensual['Ventas'].sum()
```

```python
total_ventas
```

```python
print(f"El total de ventas es: {total_ventas}")
```

```python
ventas_mensual['crecimiento%']=(ventas_mensual['Ventas'].pct_change() * 100).round(2)
```
